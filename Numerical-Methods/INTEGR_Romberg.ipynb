Metodo de integracion numerica de Romberg. Es muy similar al metodo de derivacion de Richardson pero para integracion (esta en los apuntes de MEN, p14 t5).

$R(0,0) = \frac{h_0}{2}[f(a)+f(b)]$, donde $h_n = \frac{(b-a)}{2^n}$ y donde $a,b$ son los límites de integración.

$R(K,0) = \frac{1}{2}\left[ R(K-1,0) + h_{k-1} \sum_{i=1}^{2^{k}-1}f\left( a + \left[ i - \frac{1}{2} \right]h_{k-1} \right) \right]$

$R(J,K) = \frac{4^K R(J,K-1) - R(J-1,K-1)}{4^K - 1}$

Aquí lo primero es crear la función $h_n$: 

```python
def hn(a,b,n):
    # a = limite de integracion inferior
    # b = limite de integracion superior
    # n = numero de intervalos

    valor = (b-a) / (2**n)
    return valor 
```

```python
def R_k(func,a,b,k):
    # a = limite de integracion inferior
    # b = limite de integracion superior
    # func = funcion a integrar
    # k = orden/iteraciones para el calculo de la integral
    # n = numero de intervalos 

    r_k = np.zeros((k+1,k+1)) # Genero una matriz de ceros 

    r_k[0][0] = 0.5 * (hn(a,b,0) * (func(a) + func(b))) # R(0,0)

    for m in range (1,k+1,1):
        sum = 0
        for l in range (1,2**(m-1)+1,1): # Añado el +1 al 2**(m-1) para que la suma llegue hasta 2**(m-1)
            sum = sum + func(a+(l-0.5)*hn(a,b,m-1))

        r_k[m][0] = 0.5 * ( r_k[m-1][0] + hn(a,b,m-1) * sum ) # R(K,0)

    for j in range (1,k+1,1):
        for i in range (j,k+1,1): # Calculo por columnas: columna 1 y paso por todas las filas; columna 2 y todas las filas; etc
            r_k[i][j] = (((4**j)*r_k[i][j-1]) - r_k[i-1][j-1]) / ((4**j) - 1)
    
    return r_k[k][k]
```

Programa principal para resolver $\int_{3}^{5} \log(x) dx$: 

```python
# Programa principal ------------------------------------------------------------

def f(x):
    valor = np.log(x)
    return valor

k = 3
a = 3
b = 5

solucion = R_k(f,a,b,k) # Valor aproximado de la integral 

print('El valor de la integral de f(x) es: ', solucion, '.')
```
