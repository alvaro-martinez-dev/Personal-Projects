La idea es generar numeros aleatorios uniformemente en el rango de integracion y hacer una especie de promedio (apuntes MEN t5). Matemáticamente es:

$\theta = (b-a)\left[\frac{1}{N} \sum_{i=1}^{N}g(x_i) \right] = (b-a) \, \rm{avg}[g(x)]$

$\sigma(\theta) = \frac{b-a}{\sqrt{N}} \, \sqrt{\frac{1}{N}\,\sum_{i=1}^{N}\left[g(x_i)\right]^2 - \left[\frac{1}{N}\sum_{i=1}^{N}g(x_i)\right]^2} = \frac{b-a}{\sqrt{N}} \, \sqrt{\rm{avg}[(g(x))^2] - (\rm{avg}[g(x)])^2}$

¡¡ SOLO PARA EL CASO PARA FUNCIONES 1D: $f(x)$ !! PARA EL CASO GENERAL USAR EL OTRO FICHERO

```python
def IntMC(func,a,b,N):
    # func = funcion a integrar
    # a = limite de integracion inferior
    # b = limite de integracion superior
    # N = numero de datos generados

    x_rand = np.random.uniform(a, b, N) # Vector de numeros aleatorios entre a y b

    avg_g = np.mean(func(x_rand)) # Usado para calcular integral y su error
    avg_g2 = np.mean((func(x_rand))**2) # # Usado luego para el error de la integral

    integral = float((b-a)*avg_g) # Valor aproximado de la integral

    errintegr = float(((b-a)/np.sqrt(N)) * np.sqrt(avg_g2 - avg_g**2))


    return integral,errintegr
```

El problema a resolver (por ejemplo) es hallar el valor de $\int_{3}^{5} \log (x) dx$ : 

```python
def f(x):
    valor = np.log(x)
    return valor

sol = IntMC(f,3,5,10000)

print("La solucion es (Integral, Incertidumbre): ",sol, '.')
```
