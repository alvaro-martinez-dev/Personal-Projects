La idea es ir reduciendo la matriz hasta llegar a una matriz $2 \times 2$. El concepto básico es hacer lo siguiente: 

1) Tenemos una matriz cuadrada $N \times N$
2) Hacemos $det_{N} = \sum_{i=0,j=0}^{i=N-1,j=N-1}c_{ij}$, con $c_{ij} = (-1)^{i+j} det_{N-1}$.

Resumen: hay que ir reduciendo la matriz hasta una $2 \times 2$.

Funcion factorial:

```python
def fact(N):

    Nfact = 1

    if N>1:
        for k in range(1,N+1,1): #El final(N+1) no esta incluido
            Nfact = Nfact * k

    return Nfact
```
