```python
import pandas as pd              # Para análisis de datos
import numpy as np               # Para cálculos numéricos
import matplotlib.pyplot as plt  # Para gráficos
import seaborn as sns            # Para gráficos bonitos
```

La idea es generar numeros aleatorios uniformemente en el rango de integracion y hacer una especie de promedio (apuntes MEN t5). Matemáticamente es:

$\theta = V \left[\frac{1}{N} \sum_{i=1}^{N}g(\vec{x_i}) \right] = V \, \rm{avg}[g(\vec{x})]$, donde $V$ es el volumen de integración. 

$\sigma(\theta) = \frac{V}{\sqrt{N}} \, \sqrt{\frac{1}{N}\,\sum_{i=1}^{N}\left[g(\vec{x_i})\right]^2 - \left[\frac{1}{N}\sum_{i=1}^{N}g(\vec{x_i})\right]^2} = \frac{V}{\sqrt{N}} \, \sqrt{\rm{avg}[(g(\vec{x}))^2] - (\rm{avg}[g(\vec{x})])^2}$.

Necesitamos conocer $g(\vec{x})$, los límites inferiores de integración en forma vector, $\vec{a}$, los límites superiores $\vec{b}$, y el número de puntos generados $N$.

```python
def IntMC(func,a,b,N):
    # func = funcion a integrar
    # a = vector de limites inferiores (dimension de f)
    # b = vector de limites superiores (dimension de f)
    # N = numero de puntos por dimension a generar

    dim = len(a)
    dim_comprob = len(b)

    if dim == dim_comprob:
        
        X_rand = np.zeros((dim,N)) # dim filas y N columnas

        for k in range (0,dim,1):
            X_rand[k] = np.random.uniform(a[k], b[k], N) # Relleno matriz de numeros aleatorios entre a[k] y b[k]

        avg_g = np.mean(func(*X_rand)) #  *X_rand = (X_rand[0],X_rand[1],...)
        avg_g2 = np.mean((func(*X_rand))**2)

        V = 1

        for k in range (0,dim,1):
            V = V*(b[k]-a[k])

        integral = float(V*avg_g) # Valor aproximado de la integral

        errintegr = float((V/np.sqrt(N)) * np.sqrt(avg_g2 - avg_g**2))

    else:
        print('Error: a y b tienen dimensiones diferentes.')
        return

    return integral,errintegr
```

Ejemplo $g(x,y,z) = e^{-(x^2+y^2+z^2)}$ entre $0$ y $1$. Es decir: $\int_{0}^{1}\int_{0}^{1}\int_{0}^{1}g(x,y,z)dx dy dz$ . 

```python
def func(x,y,z):
    val = np.e**(-(x**2 + y**2 + z**2))
    return val

a = np.array([0,0,0])
b = np.array([1,1,1])
N = 10000

integr,incertid = IntMC(func,a,b,N)

print('La solucion es ',integr,' + - ',incertid,'.')
```
