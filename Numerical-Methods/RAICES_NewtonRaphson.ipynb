Programa con el algoritmo de Newton-Raphson para hallar las raices de una funcion (pagina 154 Chapra). ¡¡¡OJO QUE PARA POLINOMIOS SOLO NOS DEVUELVE UN VALOR!!! 

Consiste en hacer $x_{i+1} = x_{i} - \frac{f(x_{i})}{f'(x_{i})}$ durante varias iteraciones. ¡¡¡OJO QUE LA DERIVADA NO PUEDE SER CERO!!!

Necesitamos conocer $f(x)$, su derivada $f'(x)$, un valor inicial de búsqueda $x_{ini}$, el error relativo que queremos $errel$ y el numero maximo de iteraciones $maxit$.

```python
def NewtRaph(func,dfunc,xi,errel,maxit):
    # func : funcion matematica
    # dfunc : derivada de func
    # xini : valor inicial de busqueda
    # errel : error relativo deseado
    # maxit : numero maximo de iteraciones
    # raiz : valor final
    # errapp : error relativo aproximado
    # iter : numero de iteraciones realizado

    if func==None or dfunc==None or xi==None:

        print('Error: Necesitas introducir al menos los 3 primeros elementos (func,dfunc,xini).')
        return 

    else:
        if dfunc == 0:
            print('Error: La derivada es nula.')
            return 
        else:
            if errel==None: # Si el errel introducido es vacio se asigna un valor predeterminado
                errel = 0.0001
            if maxit==None: # Si maxit introducido es vacio se asigna un valor predeterminado
                maxit = 50

            iter = 0
            errapp = 100 # Por poner algo

            while iter < maxit and errapp > errel and dfunc(xi)!=0:
                xi0 = xi # Almaceno el valor de xini incial de la iteracion
                xi = xi - func(xi)/dfunc(xi)
                iter = iter + 1
                errapp = abs((xi-xi0)/xi)*100

            raiz = xi

            return raiz,errapp,iter
```

El problema a resolver en este caso es encontrar la raíz de $f(x) = e^{-x} - x$, es decir $x_{rz}$ tal que $f(x_{rz})=0$. La derivada de esta función es $f'(x) = -e^{-x} -1$. Iniciamos la busqueda en $x_{ini} = 0$. 

```python
def f(x): # Esta funcion puede ser la que me de la gana
    fx = np.e ** (-x) - x
    return fx

def fprime(x): # Derivada de la funcion anterior
    fpx = - np.e ** (-x) -1
    return fpx

solucion = NewtRaph(f,fprime,0,None,3)

print('( Raiz, Error Relativo(%), Numero iteraciones )  = ', solucion)
```
