Programa con el algoritmo de Punto Fijo para hallar las raices de una funcion (pagina 146 Chapra).
¡¡OJO QUE PARA POLINOMIOS SOLO NOS DEVUELVE UN VALOR!!

El algoritmo consiste en despejar una $x$ de la función $f(x) = 0$ con tal de obtener $x = g(x)$. Entonces el algoritmo es básicamente hacer $x_{k+1} = g(x_k)$. 

Necesitamos conocer $f(x)$, un valor inicial de búsqueda $x_{ini}$, el error relativo deseado $errel$ y el número de iteraciones máximo que queremos $maxit$. 

```python
def PuntoFijo(gx,xi,errel,maxit):
    # gx = g(x)
    # xi = punto inicial de busqueda
    # errel = error relativo deseado
    # maxit = numero maximo de iteraciones
    # raiz = valor final de xi
    # errapp = error relativo aproximado
    # iter = numero de iteraciones realizadas

    if gx == None or xi == None:
        print('Debes introducir la funcion g(x) y un valor inicial de busqueda xi.')
        return 
    else:
        if errel == None:
            errel = 0.001
        if maxit == None:
            maxit = 50
        
        iter = 0
        errapp = 100 # Por poner algo

        while iter < maxit and errapp > errel:
            xi0 = xi
            xi = gx(xi)
            iter = iter +1
            errapp = abs((xi-xi0)/xi) * 100

        raiz = xi    

        return raiz, errapp, iter
```

Estudiaremos la raiz de $f(x) = x - e^{-x}$ (por ejemplo) haciendo $x = g(x) = e^{-x}$.

```python
def gx(x):
    valor = np.e ** (-x)

    return valor

solucion = PuntoFijo(gx,0,None,None)

print('( Raiz, Error Relativo(%), Numero iteraciones )  = ', solucion)
```
