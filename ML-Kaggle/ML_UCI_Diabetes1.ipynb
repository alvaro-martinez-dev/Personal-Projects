## Gestión inicial de los datos

```python
data.columns
```

Las features que tenemos son:
- Diabetes_binary: 0 = no diabetes 1 = prediabetes or diabetes
- HighBP: 0 = no high BP 1 = high BP
- HighChol: 0 = no high cholesterol 1 = high cholesterol
- CholCheck: 0 = no cholesterol check in 5 years 1 = yes cholesterol check in 5 years
- BMI
- Smoker: Have you smoked at least 100 cigarettes in your entire life? [Note: 5 packs = 100 cigarettes] 0 = no 1 = yes
- Stroke: (Ever told) you had a stroke. 0 = no 1 = yes
- HeartDiseaseorAttack: coronary heart disease (CHD) or myocardial infarction (MI) 0 = no 1 = yes
- PhysActivity: physical activity in past 30 days - not including job 0 = no 1 = yes
- Fruits: Consume Fruit 1 or more times per day 0 = no 1 = yes
- Veggies: Consume Vegetables 1 or more times per day 0 = no 1 = yes
- HvyAlcoholConsump: (adult men >=14 drinks per week and adult women>=7 drinks per week) 0 = no 1 = yes
- AnyHealthcare: Have any kind of health care coverage, including health insurance, prepaid plans such as HMO, etc. 0 = no 1 = yes
- NoDocbcCost: Was there a time in the past 12 months when you needed to see a doctor but could not because of cost? 0 = no 1 = yes
- GenHlth: Would you say that in general your health is: scale 1-5 1 = excellent 2 = very good 3 = good 4 = fair 5 = poor
- MenHlth: days of poor mental health scale 1-30 days
- PhysHlth: physical illness or injury days in past 30 days scale 1-30
- DiffWalk: Do you have serious difficulty walking or climbing stairs? 0 = no 1 = yes
- Sex: 0 = female 1 = male
- Age: 13-level age category (_AGEG5YR see codebook) 1 = 18-24 9 = 60-64 13 = 80 or older
- Education: Education level (EDUCA see codebook) scale 1-6 1 = Never attended school or only kindergarten...
- Income: Income scale (INCOME2 see codebook) scale 1-8 1 = less than $10,000 5 = less than $35,000 8 = $75,000 or more

Tal vez podríamos establecer que aquellos valores con $\pm 0.1$ en el heatmap no tienen relevancia a la hora de determinar si hay o no diabetes. Así que por hacer el modelo más sencillo podemos probar a quitarlo.

```python
data_red = data[['Diabetes_binary', 'HighBP', 'HighChol', 'BMI',
       'Stroke', 'HeartDiseaseorAttack', 'PhysActivity', 'GenHlth', 
       'PhysHlth', 'DiffWalk', 'Age', 'Education','Income']]
```

```python
data_red.describe()
```

Vamos a normalizar el BMI y PhysActivity entre 0 y 1.

```python
from sklearn.preprocessing import MinMaxScaler #Para normalizar los datos entre 0 y 1

# Inicializa el scaler
scaler = MinMaxScaler()

# Escala las columnas y reemplaza en el DataFrame
data_red['BMI'] = scaler.fit_transform(data_red[['BMI']])
data_red['PhysHlth'] = scaler.fit_transform(data_red[['PhysHlth']])
```

## Modelo ML

```python
X_data = data_red[['HighBP', 'HighChol', 'BMI',
       'Stroke', 'HeartDiseaseorAttack', 'PhysActivity', 'GenHlth', 
       'PhysHlth', 'DiffWalk', 'Age', 'Education','Income']]
y_data = data_red.Diabetes_binary
```

```python
from sklearn.model_selection import train_test_split

# Dividimos el train original en train + test
X_train, X_test, y_train, y_test = train_test_split(X_data, y_data, test_size=0.2, random_state=42)
```

```python
# Dividimos el train original en train + valid
X_train_new, X_valid, y_train_new, y_valid = train_test_split(X_train, y_train, test_size=0.2, random_state=42)
```

```python
from xgboost import XGBClassifier

diabetes_xgmodel = XGBClassifier(
    n_estimators=1000,        # número de árboles
    learning_rate=0.05,      # tamaño de los pasos de boosting
    max_depth=8,             # profundidad máxima de cada árbol
    subsample=0.9,           # fracción de datos por árbol
    colsample_bytree=0.7,    # fracción de features por árbol
    use_label_encoder=False,
    eval_metric="logloss",
    early_stopping_rounds=50, 
    verbose=False,
    objective="binary:logistic",
    min_child_weight= 7)

diabetes_xgmodel.fit(X_train_new, y_train_new, eval_set=[(X_valid, y_valid)])
```

```python
predictions_xgmodel = diabetes_xgmodel.predict(X_test)

from sklearn.metrics import accuracy_score

# Suponiendo que y_test y predictions son arrays/listas
accuracy_xgmodel = accuracy_score(y_test, predictions_xgmodel) * 100
print(f"Porcentaje de aciertos: {accuracy_xgmodel:.2f}%")
```

```python
from xgboost import plot_importance

plot_importance(diabetes_xgmodel, importance_type='gain')
plt.grid(False)
plt.show()
```

```python
from sklearn.metrics import roc_curve, roc_auc_score

# Predecir probabilidades de la clase positiva
y_prob = diabetes_xgmodel.predict_proba(X_data)[:, 1]

# Calcular curva ROC
fpr, tpr, thresholds = roc_curve(y_data, y_prob)
auc_score = roc_auc_score(y_data, y_prob)

# Graficar
plt.plot(fpr, tpr, label=f"AUC = {auc_score:.2f}")
plt.plot([0, 1], [0, 1], linestyle='--', color='red')  # línea diagonal
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()
plt.grid(False)  # quita la rejilla
plt.show()
```
